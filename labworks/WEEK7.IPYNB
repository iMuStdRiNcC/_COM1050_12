```python
#W7A1
def binary_search(arr,value):
    left , right = 0 , len(arr) -1 
    
    while True:
        mid = (left+right)//2
        if arr[mid] < value:
            left += 1
        elif arr[mid] > value:
            right -= 1
        elif arr[mid] == value:
            return mid
        else:
            return -1
        
    
        
seq = list(map(int,input().split()))
number = int(input())
print(binary_search(seq,number))
```

```python
#W7A2
def count_occ(seq,value):
    count = 0
    for i in seq:
        if i == value:
            count+= 1
    return count 

seq = list(map(int,input().split()))
number = int(input())
print(count_occ(seq,number))
```

```python
#W7A3
def bubble_sort(seq):
    n = len(seq)
    for i in range(n):
        swapped = False
        for j in range(n-i-1):
            if seq[j] > seq[j+1]:
                seq[j],seq[j+1] = seq[j+1],seq[j]
                swapped = True
        if not swapped:
            break 
    return seq

seq = list(map(int,input().split()))

print(bubble_sort(seq))
```

```python
#W7A5
def count_pairs_with_sum(nums, X):
    count = 0
    seen_numbers = {}

    for num in nums:
        complement = X - num
        
        if complement in seen_numbers:
            count += seen_numbers[complement]
            
        seen_numbers[num] = seen_numbers.get(num, 0) + 1
        
    return count


numbers = [2, 3, 2, 3, 1, 4, 5, 6, -1, 3, 2, 3]
X = 5

result = count_pairs_with_sum(numbers, X)
print(f"{result}")
```

```python
#W7A7
a = list(map(int, input().strip()[1:-1].split(',')))
n = len(a)
tail = [min(a) - 1] + [max(a) + 1] * n
ans = 0
for i in range(n):
    l, r = 0, i
    while l <= r:
        mid = l + r >> 1
        if tail[mid] < a[i]:
            l = mid + 1
        else:
            r = mid - 1
    ans = max(ans, r + 1)
    tail[r + 1] = min(tail[r + 1], a[i])
print(ans)
```

```python
#W7A8

a = list(list(int(j) for j in i.split(',')) for i in ''.join(input().strip()[2:-2].split()).split('],['))
a = sorted(list([i[1] - i[0] + 1, i[0], i[1]] for i in a))

q = list(map(int, input().strip()[1:-1].split(',')))
q = sorted(list([q[i], i] for i in range(len(q))))

n = len(q)
ans = [-1] * n

def bin_l(x):
    l, r = 0, n - 1
    while l <= r:
        mid = l + r >> 1
        if q[mid][0] >= x:
            r = mid - 1
        else:
            l = mid + 1
    return l if l < n else -1

def bin_r(x):
    l, r = 0, n - 1
    while l <= r:
        mid = l + r >> 1
        if q[mid][0] <= x:
            l = mid + 1
        else:
            r = mid - 1
    return r if r >= 0 else -1

par = [i for i in range(n + 1)]

def find(i):
    if par[i] == i:
        return i
    par[i] = find(par[i])
    return par[i]

for [len, i, j] in a:
    if i > q[-1][0] or j < q[0][0]:
        continue
    l = bin_l(i)
    r = bin_r(j)
    assert l != -1 and r != -1
    cur = find(l)
    while cur <= r:
        ans[q[cur][1]] = len
        par[cur] = find(cur + 1)
        cur = find(cur)

print(ans)
```

```python
#W7A9

a = list(map(int, input().strip()[1:-1].split(',')))

n = len(a)
ans = [0] * n

b = [[a[i], i] for i in range(n)]

st = [[0, n - 1, 0]]

while st:
    l, r, fl = st.pop()
    if l >= r:
        continue
    mid = l + r >> 1
    if fl:
        i, j = l, mid + 1
        tmp = []
        while i <= mid and j <= r:
            if b[j][0] < b[i][0]:
                tmp += [b[j]]
                j += 1
            else:
                ans[b[i][1]] += j - mid - 1
                tmp += [b[i]]
                i += 1
        while i <= mid:
            ans[b[i][1]] += j - mid - 1
            tmp += [b[i]]
            i += 1
        while j <= r:
            tmp += [b[j]]
            j += 1
        for i in range(r - l + 1):
            b[l + i] = tmp[i]
    else:
        st += [[l, r, 1], [l, mid, 0], [mid + 1, r, 0]]

print(ans)
```

```python
#W7A10

a = list(map(int, input().strip()[1:-1].split(',')))
m = int(input())

n = len(a)
pre = [0] * n
for i in range(n):
    pre[i] = ((pre[i - 1] + a[i] % m) + m) % m if i else ((a[i] % m) + m) % m
st = [[0, n - 1, 0]]

ans = max(pre)

while st:
    l, r, fl = st.pop()
    if l >= r:
        continue
    mid = l + r >> 1
    if fl:
        i, j = l, mid + 1
        tmp = []
        while i <= mid and j <= r:
            if pre[i] <= pre[j]:
                tmp += [pre[i]]
                i += 1
            else:
                ans = max(ans, pre[j] - pre[i] + m)
                tmp += [pre[j]]
                j += 1
        while i <= mid:
            tmp += [pre[i]]
            i += 1
        while j <= r:
            tmp += [pre[j]]
            j += 1
        for i in range(r - l + 1):
            pre[l + i] = tmp[i]
    else:
        st += [[l, r, 1], [l, mid, 0], [mid + 1, r, 0]]
print(ans)
```
